```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
x=np.array([1,2,3,4,5])
y=np.array([20,40,50,65,80])
x_mean = np.mean(x)
y_mean = np.mean(y)
print("x mean:", x_mean)
print("y mean:",y_mean)
```

```python
b1=np.sum((x-x_mean)*(y-y_mean))/np.sum((x-x_mean)**2)
print("slope b1:",b1)
```

```python
b0=y_mean-b1*x_mean
print("intercept b0:",b0)
```

```python
y_predicted = b0 + b1*x
print("y predicted values:", y_predicted)
```

```python
MAE = mean_absolute_error(y, y_predicted)
print("MAE:",MAE)
```

```python
MSE = mean_squared_error(y, y_predicted)
print("MSE:",MSE)
```

```python
RMSE = np.sqrt(MSE)
print("RMSE:",RMSE)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Actual Values')
plt.plot(x, y_predicted, color='red', linestyle='-', label='Predicted Line')
plt.title('Actual vs. Predicted Values')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
